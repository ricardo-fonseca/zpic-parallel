## Custom plot

```python
import numpy as np
import zdf
import matplotlib.pyplot as plt
import matplotlib.colors as colors

iter = 2300
# iter = 4000

title ="Charge density and Laser field"

# charge_map = "GnBu"
# charge_map = "Greys"
charge_map = "inferno_r"

charge_max = 12

laser_map = "viridis_r"
# laser_map ="plasma"


def r_mirror( data ):
    mirror = np.zeros( [ 2 * data.shape[0] - 1, data.shape[1] ], dtype=np.float32 )
    mirror[data.shape[0]-1:,:] = data
    mirror[0:data.shape[0],:] = np.flip( data, axis = 0 )
    return mirror


(charge, info) = zdf.read("CHARGE/electrons/electrons-ρ0-{:06d}.zdf".format(iter))

# Mirror range
range = [
    [info.grid.axis[0].min, info.grid.axis[0].max],
    [-info.grid.axis[1].max, info.grid.axis[1].max]
]

charge = r_mirror( np.abs(charge) )

plt.imshow( charge, interpolation = 'nearest', origin = 'lower',aspect = 'auto',
           extent = ( range[0][0], range[0][1], range[1][0], range[1][1] ),
           cmap = charge_map, vmax = charge_max )
plt.colorbar().set_label("$n / n_0$")

plt.title(title + " \n $ t = " + "{:.3g}".format(info.iteration.t) + "\\, [\\mathrm{1/\\omega_n}]$")
plt.xlabel("$ z \\, [\\mathrm{c/\\omega_n}]$")
plt.ylabel("$ r \\, [\\mathrm{c/\\omega_n}]$")


(laser, info) = zdf.read("EMF/E1r-{:06d}.zdf".format(iter))

laser = r_mirror( np.abs( laser ) )

alpha = np.clip( 5 * laser / np.max( laser ), a_min = 0, a_max = 1 )

plt.imshow( laser, interpolation = 'nearest', origin = 'lower',aspect = 'auto',
           extent = ( range[0][0], range[0][1], range[1][0], range[1][1] ),
           cmap = laser_map, alpha = alpha)

plt.axhline(y=0, color='k', linestyle='-.', linewidth = 1 )

# plt.savefig("q3d.pdf", format="pdf", bbox_inches="tight")

plt.show()
```
